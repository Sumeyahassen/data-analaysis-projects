```python
import pandas as pd
import numpy as np

# lode telecome custemer data
telecom_customer_df =pd.read_csv('../data/processed/merged_telecom_data.csv')
print(f'lode data is {telecom_customer_df.shape}')
x=telecom_customer_df.isnull().sum().sum()
print(x)
```

```python
# to see the missing data
missing_data=telecom_customer_df.isnull().sum()
print("\nmiss data contane columens:\n", missing_data[missing_data > 0])
telecom_customer_df.info()
```

```python
# 1.  Numerical Columns using Median to replase
numerical_columns = telecom_customer_df.select_dtypes(include=np.number).columns
for col in numerical_columns:
    if telecom_customer_df[col].isnull().sum() > 0:
        median_val = telecom_customer_df[col].median()
        telecom_customer_df[col] = telecom_customer_df[col].fillna(median_val)

# 2. Categorical/Text Columnsreplase by 'Unknown' 
categorical_columns = telecom_customer_df.select_dtypes(include=['object', 'str']).columns
for col in categorical_columns:
    if telecom_customer_df[col].isnull().sum() > 0:
        telecom_customer_df[col] = telecom_customer_df[col].fillna('Unknown')

# 3. to check null value are correctely to replase or note
remaining_null = telecom_customer_df.isnull().sum().sum()
print(f'\nCurrently null value amount: {remaining_null}')
```

```python
# Save Clean Data
telecom_customer_df.to_csv('../data/processed/merged_telecom_data.csv', index=False)
print("save clean data successfully")
```

```python
# 2. textual and catagorical data conver One-Hot Encoding

# ml algorisems are not understand the textual data so conver in to number
telecom_customer_encoded_df = pd.get_dummies(telecom_customer_df, drop_first=True)

print("converted data shape:", telecom_customer_encoded_df.shape)
telecom_customer_encoded_df.head()
```
