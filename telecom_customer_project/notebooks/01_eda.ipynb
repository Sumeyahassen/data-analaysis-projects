```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# load data of cliant and recorde
client_df = pd.read_csv('../data/raw/Client.csv')
record_df = pd.read_csv('../data/raw/Record.csv')
```

```python
#  to see the colomens and sample data
print( "cliant columens",client_df.columns)
print('recordes columens',record_df.columns)
```

```python
# to see the the first 5 rows
print(client_df.head)
```

```python
print(record_df.head)
```

```python
# for both data there is  Cliant id columen so to merge it
telecom_customer_df = pd.merge(client_df,record_df,on='Customer_ID',how='inner')
# MERGE  datas
print('merge datas',telecom_customer_df.shape,telecom_customer_df.head())
print()
```

```python
# to see the data information of telecom_customer_df
telecom_customer_df.info()
```

```python
# to see the chunc cliant as persentile
churn_percentage =telecom_customer_df['churn'].value_counts(normalize=True)*100
print(f'churn cliant peresentage rate \n',churn_percentage)
```

```python
plt.figure(figsize=(6,4))
sns.countplot(data=telecom_customer_df,x='churn')
plt.title('Customer Churn Distribution')
plt.savefig('../outputs/figures/churn_distribution.png')
plt.show
```

```python
output_path = '../data/processed/merged_telecom_data.csv'
telecom_customer_df.to_csv(output_path, index=False)

print(f"to save successfully: {output_path}")
```
